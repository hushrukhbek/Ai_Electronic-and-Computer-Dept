```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract coefficients
a, b, c = mymodel.coefficients

# Prediction function
def f(x_value):
    return mymodel(x_value)

# Print equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Generate x-values
myline = np.linspace(1, 22, 100)

# 17:00 dagi prediction
predicted_speed = f(17)

# 1. Print predicted speed
print(f"Predicted speed at 17:00: {predicted_speed:.2f}")

# Graph
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')

# 2. Display predicted value on graph
plt.scatter(17, predicted_speed, color='blue', s=100,
            label=f'17:00 prediction = {predicted_speed:.2f}')

plt.title("Polynomial Regression (degree=2)")
plt.xlabel("Time")
plt.ylabel("Speed")
plt.legend()
plt.show()
```
